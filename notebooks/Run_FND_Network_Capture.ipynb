## View generated figures
After running the notebook, all six figures are displayed below.


```python
from pathlib import Path
from IPython.display import display, Image, Markdown

for i in range(1, 7):
    p = Path(f"figures/Figure{i}.png")
    if p.exists():
        display(Markdown(f"### Figure {i}"))
        display(Image(filename=str(p)))
    else:
        print(f"Missing: {p}")
```

## View numerical results
Preview the generated CSV files and numerical verification report.


```python
import pandas as pd
import json
from IPython.display import display

for p in sorted(Path("results").glob("*.csv")):
    display(Markdown(f"### {p.name}"))
    display(pd.read_csv(p))
verification = Path("results/numerical_verification.json")
if verification.exists():
    display(Markdown("### Numerical verification"))
    print(json.dumps(json.loads(verification.read_text()), indent=2))
```

## Download all figures and results
Run this cell to create and download a ZIP containing the generated figures and numerical results.


```python
import shutil
from google.colab import files
archive = shutil.make_archive("FND_Network_Capture_results", "zip", root_dir=".", base_dir="figures")
# Include the results folder in the same archive.
import zipfile
with zipfile.ZipFile(archive, "a") as bundle:
    for p in sorted(Path("results").rglob("*")):
        if p.is_file():
            bundle.write(p, arcname=str(p))
files.download(archive)
```
